### 3. Generating a sinusoid

3.1. Create a sinusoid of the same duration than the reference signal and same frequency than the fundamental frequency. Plot it along with the audio signal of the reference recording. Plot the same time segment that 2.3. Note that the sampling rate should be the same for both signals. Try to find the values of Amplitude and initial Phase that make the two signal segments plotted to match as close as possible (do not attempt to imitate the whole signal). Matching the phase is a bit tricky.



```python
import numpy as np

# Duración total basada en la señal original
num_samples = len(ref)
duration = num_samples / fs
t = np.linspace(0, duration, num_samples, endpoint=False)

# Parámetros ajustados al período medido (~2.1 ms -> f0 ≈ 476 Hz)
f0 = 476.0      # Frecuencia fundamental medida
A = 0.09        # Amplitud observada en la zona estable
phi = -np.pi / 2 # Fase inicial en radianes para alinear los picos

# Generación de la sinusoide pura
synthesis_sin = A * np.sin(2 * np.pi * f0 * t + phi)

# Graficar ambas señales superpuestas en la misma ventana de 3 períodos
plot_signals([ref, synthesis_sin], fs, t_start=t_start, t_end=t_end)
```

3.2. What are the main differences between the two signals.

La señal sintetizada coincide bien en frecuencia y amplitud con el componente fundamental de la señal de referencia. Sin embargo, difieren en el timbre porque el ukelele real contiene múltiples armónicos y un perfil de ataque/decaimiento dinámico que la sinusoide pura no posee.
